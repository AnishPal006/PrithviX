```python
from pathlib import Path

PROJECT_DIR = Path("/Users/rajdeepmandal/Desktop/PrithviX")

DATA_DIR = PROJECT_DIR / "Data"

BOUNDARY_DIR = DATA_DIR / "Boundaries"
RAINFALL_DIR = DATA_DIR / "Rainfall"
DEM_DIR = DATA_DIR / "DEM"
NDVI_DIR = DATA_DIR / "NDVI"
LANDCOVER_DIR = DATA_DIR / "LandCover"
SOIL_DIR = DATA_DIR / "Soil"
POPULATION_DIR = DATA_DIR / "Population"
LANDSLIDE_DIR = DATA_DIR / "LandslideInventory"
OSM_DIR = DATA_DIR / "OpenStreetMap"

print(PROJECT_DIR)
```

```python
# ==========================================
# Cell 5 - Validate All Dataset Files
# ==========================================

print("=" * 80)
print("VALIDATING ALL DATASETS")
print("=" * 80)

# ---------- CSV Files ----------

print("\n📄 Rainfall CSV Files")
print("-" * 60)

for file in sorted(RAINFALL_DIR.glob("*.csv")):

    try:
        # NASA POWER files always have the CSV header after "-END HEADER-"
        with open(file, "r", encoding="utf-8") as f:
            lines = f.readlines()

        header_idx = None
        for i, line in enumerate(lines):
            if line.strip() == "-END HEADER-":
                header_idx = i + 1   # Next line is the CSV header
                break

        if header_idx is None:
            raise ValueError("Could not find '-END HEADER-' in the file.")

        df = pd.read_csv(file, skiprows=header_idx)

        print(f"✅ {file.name}")
        print(f"   Rows    : {df.shape[0]}")
        print(f"   Columns : {df.shape[1]}")

    except Exception as e:
        print(f"❌ {file.name}")
        print(e)

# ---------- Boundary ----------

print("\n🗺 Boundary")
print("-" * 60)

for file in BOUNDARY_DIR.glob("*"):

    try:

        gdf = gpd.read_file(file)

        print(f"✅ {file.name}")
        print(f"   Features : {len(gdf)}")
        print(f"   CRS      : {gdf.crs}")

    except Exception as e:

        print(f"❌ {file.name}")
        print(e)


# ---------- DEM ----------

print("\n🏔 DEM")
print("-" * 60)

for file in DEM_DIR.glob("*.tif"):

    try:

        src = rasterio.open(file)

        print(f"✅ {file.name}")
        print(f"   Width  : {src.width}")
        print(f"   Height : {src.height}")
        print(f"   CRS    : {src.crs}")

    except Exception as e:

        print(f"❌ {file.name}")
        print(e)


# ---------- NDVI ----------

print("\n🌿 NDVI")
print("-" * 60)

for file in NDVI_DIR.glob("*.tif"):

    try:

        src = rasterio.open(file)

        print(f"✅ {file.name}")
        print(f"   Width  : {src.width}")
        print(f"   Height : {src.height}")
        print(f"   CRS    : {src.crs}")

    except Exception as e:

        print(f"❌ {file.name}")
        print(e)


# ---------- Land Cover ----------

print("\n🟩 Land Cover")
print("-" * 60)

for file in LANDCOVER_DIR.glob("*.tif"):

    try:

        src = rasterio.open(file)

        print(f"✅ {file.name}")
        print(f"   Width  : {src.width}")
        print(f"   Height : {src.height}")
        print(f"   CRS    : {src.crs}")

    except Exception as e:

        print(f"❌ {file.name}")
        print(e)


# ---------- Soil ----------

print("\n🌍 Soil")
print("-" * 60)

soil_files = list(SOIL_DIR.glob("*.tif"))

if not soil_files:
    print("❌ No soil raster found.")
else:
    for file in soil_files:
        ...

    try:

        src = rasterio.open(file)

        print(f"✅ {file.name}")
        print(f"   Width  : {src.width}")
        print(f"   Height : {src.height}")
        print(f"   CRS    : {src.crs}")

    except Exception as e:

        print(f"❌ {file.name}")
        print(e)


# ---------- Population ----------

print("\n👥 Population")
print("-" * 60)

for file in POPULATION_DIR.glob("*.tif"):

    try:

        src = rasterio.open(file)

        print(f"✅ {file.name}")
        print(f"   Width  : {src.width}")
        print(f"   Height : {src.height}")
        print(f"   CRS    : {src.crs}")

    except Exception as e:

        print(f"❌ {file.name}")
        print(e)


# ---------- Landslide ----------

print("\n⚠ Historical Landslides")
print("-"*60)

for file in LANDSLIDE_DIR.glob("*.csv"):

    try:

        # Find the actual header row
        header_line = 0

        with open(file, "r") as f:
            lines = f.readlines()

            for i, line in enumerate(lines):

                if "SL.NO" in line.upper():
                    header_line = i
                    break

        df = pd.read_csv(file, skiprows=header_line)

        print(f"✅ {file.name}")
        print(f"   Rows : {len(df)}")
        print(f"   Columns : {len(df.columns)}")

    except Exception as e:

        print(f"❌ {file.name}")
        print(e)

# ---------- OSM ----------

print("\n🛣 OpenStreetMap")
print("-" * 60)

for file in OSM_DIR.glob("*.gpkg"):

    try:

        gdf = gpd.read_file(file)

        print(f"✅ {file.name}")
        print(f"   Features : {len(gdf)}")
        print(f"   CRS      : {gdf.crs}")

    except Exception as e:

        print(f"❌ {file.name}")
        print(e)

print("\n" + "=" * 80)
print("VALIDATION COMPLETED")
print("=" * 80)
```

```python
# ==========================================
# Cell 6 - Inspect Dataset Contents
# ==========================================

print("=" * 80)
print("DATASET CONTENT INSPECTION")
print("=" * 80)

# ------------------------------------------------
# Rainfall Sample
# ------------------------------------------------
print("\n🌧 Rainfall Sample")
print("-" * 60)

sample_rain = sorted(RAINFALL_DIR.glob("*.csv"))[0]

# Read NASA POWER CSV
with open(sample_rain, "r", encoding="utf-8") as f:
    lines = f.readlines()

header_idx = None
for i, line in enumerate(lines):
    if line.strip() == "-END HEADER-":
        header_idx = i + 1
        break

rain_df = pd.read_csv(sample_rain, skiprows=header_idx)

print("Columns:")
print(list(rain_df.columns))

print("\nFirst 5 rows:")
display(rain_df.head())


# ------------------------------------------------
# Landslide Sample
import pandas as pd

landslide_file = LANDSLIDE_DIR / "landslides.csv"

# Find the row containing the actual column names
with open(landslide_file, "r", encoding="utf-8") as f:
    lines = f.readlines()

header_row = None
for i, line in enumerate(lines):
    if "Sl.No." in line or "Slide_No" in line:
        header_row = i
        break

print("Header found at row:", header_row)

# Read from the correct header
df = pd.read_csv(landslide_file, skiprows=header_row)

# Remove completely empty rows
df = df.dropna(how="all")

# Remove unnamed columns
df = df.loc[:, ~df.columns.str.contains("^Unnamed")]

# Clean column names
df.columns = df.columns.str.strip()

print(df.columns.tolist())
display(df.head())
# ------------------------------------------------
# DEM Information
# ------------------------------------------------
print("\n🏔 DEM")
print("-" * 60)

dem_file = list(DEM_DIR.glob("*.tif"))[0]

with rasterio.open(dem_file) as src:

    print("Bounds :", src.bounds)
    print("Resolution :", src.res)
    print("Bands :", src.count)


# ------------------------------------------------
# NDVI Information
# ------------------------------------------------
print("\n🌿 NDVI")
print("-" * 60)

for file in NDVI_DIR.glob("*.tif"):

    with rasterio.open(file) as src:

        print(file.name)
        print("Resolution :", src.res)
        print("Bands :", src.count)
        print()


# ------------------------------------------------
# Population
# ------------------------------------------------
print("\n👥 Population")
print("-" * 60)

for file in POPULATION_DIR.glob("*.tif"):

    with rasterio.open(file) as src:

        print(file.name)
        print("Resolution :", src.res)
        print("Bands :", src.count)
        print()


# ------------------------------------------------
# Boundary File
# ------------------------------------------------
print("\n🗺 Boundary")
print("-" * 60)

boundary_files = list(BOUNDARY_DIR.glob("*"))

if len(boundary_files) == 0:
    print("❌ No boundary file found.")
else:
    print("Boundary file(s):")
    for file in boundary_files:
        print(file.name)

        try:
            gdf = gpd.read_file(file)
            print("Columns:")
            print(list(gdf.columns))
            print("Features:", len(gdf))
            print()

        except Exception as e:
            print(e)

print("\n" + "=" * 80)
print("INSPECTION COMPLETED")
print("=" * 80)
```
